```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

import essential libraries: pandas, matplotlib, seaborn

```python
# 1. Load dataset
df = pd.read_csv("https://raw.githubusercontent.com/duochen/data-science-bootcamp/refs/heads/main/Data/health_data.csv")
print(df.head())
```

loads datset, makes computer read it, displays first five rows

```python
# 2. Summary statistics
print("Summary:")
print(df.describe())
```

computes and prints descriptive statistics for the numerical columns in the dataset

```python
# 3. Boxplot: Outliers in Infant Mortality
sns.boxplot(x=df["Infant Mortality"])
plt.title("Outliers in Infant Mortality Rate")
plt.show()
```

Generates a horizontal boxplot to analyze the distribution of the Infant Mortality rate and highlight outliers.

## 📦 Plot 1: **Boxplot – Infant Mortality Rate**

### 🔍 What It Shows:

* A **horizontal boxplot** of the `Infant Mortality` column.
* The box represents the **interquartile range (IQR)**: the middle 50% of countries.
* The **line inside the box** is the **median**.
* **Dots or points beyond the “whiskers”** indicate **outliers** — countries with unusually high infant mortality.

### ✅ Interpretation:

* If the whiskers are short and a few dots lie far to the right, some countries have **much higher infant mortality** than the average.
* In this dataset, **Nigeria** and **South Africa** likely appear as outliers due to very high rates.
* This plot helps detect **inequality in health outcomes**.

```python
# 4. Correlation Heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="viridis")
plt.title("Health Indicators Correlation Matrix")
plt.show()
```

df.corr(numeric_only=True): Calculates the statistical correlation between all numerical columns in your dataset (df). Correlation measures how strongly two variables are related to each other (ranging from -1 to 1).

sns.heatmap(...): Takes that correlation table and turns it into a color-coded grid (a heatmap).

annot=True: Prints the exact numerical correlation values right inside each grid cell so you can easily read them.

cmap="viridis": Sets the color palette of the map to "viridis" (a smooth, colorblind-friendly gradient from purple to green)

## 🔥 Plot 2: **Correlation Heatmap**

### 🔍 What It Shows:

* A **colored matrix** of pairwise correlations between numeric columns:

  * `Life Expectancy`
  * `Health Spending Per Capita`
  * `Infant Mortality`

### ✅ Interpretation:

* Values range from **–1 (perfect negative)** to **+1 (perfect positive)**.
* Example insights:

  * 🔼 **High positive correlation** (close to +1): `Health Spending` ↑ → `Life Expectancy` ↑
  * 🔽 **High negative correlation** (close to –1): `Infant Mortality` ↑ → `Life Expectancy` ↓
* Color helps: darker = stronger relationship.

### 💡 Key Insight:

* **More health spending is generally associated with higher life expectancy and lower infant mortality.**

```python
# 5. Scatter plot: Health Spending vs Life Expectancy
sns.scatterplot(x="Health Spending Per Capita", y="Life Expectancy", data=df)
plt.title("Do Richer Countries Live Longer?")
plt.show()
```

sns.scatterplot(x="Health Spending Per Capita", y="Life Expectancy", data=df)x="Health Spending Per Capita": Sets the horizontal ($x$) axis to show how much money each country spends per person on healthcare.y="Life Expectancy": Sets the vertical ($y$) axis to show the average life expectancy for each country.data=df: Tells Seaborn to pull these columns from your dataset (df). Each dot on the plot represents a single country.

## 💬 Plot 3: **Scatter Plot – Health Spending vs. Life Expectancy**

### 🔍 What It Shows:

* Each point represents a **country**.
* X-axis: `Health Spending Per Capita`
* Y-axis: `Life Expectancy`

### ✅ Interpretation:

* A **positive upward trend** means that countries that **spend more on healthcare** tend to have **higher life expectancy**.
* Example:

  * Japan and Norway likely appear toward the top-right (high spending, long lives).
  * Nigeria and India likely appear in the lower-left (low spending, shorter lives).
