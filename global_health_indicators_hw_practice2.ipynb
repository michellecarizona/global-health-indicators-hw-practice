```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/duochen/data-science-bootcamp/refs/heads/main/Data/health_data.csv")
print(df.head())
```

```python
print("Summary:")
print(df.describe())
```

```python
sns.boxplot(x=df["Infant Mortality"])
plt.title("Outliers in Infant Mortality Rate")
plt.show()
```

```python

sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Health Indicators Correlation Matrix")
plt.show()
```

changed colors to better match the topic (heat)

```python
sns.scatterplot(x="Health Spending Per Capita", y="Life Expectancy", data=df)
plt.title("Do Richer Countries Live Longer?", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Health Spending Per Capita (USD)", fontsize=12)
plt.ylabel("Life Expectancy (Years)", fontsize=12)


plt.tight_layout()
plt.show()
```

changed font size and axis labels
